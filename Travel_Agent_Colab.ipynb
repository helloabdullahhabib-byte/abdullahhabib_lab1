This notebook contains a simple rule-based travel recommendation agent.

```python
# Travel Agent - Artificial Intelligence Lab 1
# Simple rule-based travel recommendation program
# This code can be copied into Google Colab and run directly.

destinations = [
    {"name": "Hunza", "type": "mountains", "min_budget": 25000, "days": 5,
     "best_for": "nature"},
    {"name": "Murree", "type": "mountains", "min_budget": 15000, "days": 3,
     "best_for": "relaxation"},
    {"name": "Lahore", "type": "city", "min_budget": 12000, "days": 3,
     "best_for": "history"},
    {"name": "Skardu", "type": "mountains", "min_budget": 30000, "days": 6,
     "best_for": "adventure"},
    {"name": "Karachi", "type": "city", "min_budget": 14000, "days": 3,
     "best_for": "food"}
]

print("===== TravelMate: Smart Travel Agent =====")

budget = int(input("Enter your total budget in PKR: "))
days = int(input("How many days do you want to travel? "))
interest = input("What is your main interest (nature/history/adventure/relaxation/food)? ").lower()

possible = []

for place in destinations:
    if budget >= place["min_budget"] and days >= place["days"]:
        score = 0

        # Match the user's interest
        if interest == place["best_for"]:
            score += 3

        # Prefer a destination that does not use too much of the budget
        if budget <= place["min_budget"] + 10000:
            score += 1

        # Prefer a destination close to the requested number of days
        if days <= place["days"] + 2:
            score += 1

        possible.append((score, place))

if not possible:
    print("\nSorry, I could not find a destination matching your budget and days.")
else:
    possible.sort(key=lambda x: x[0], reverse=True)
    best_score, best = possible[0]

    print("\n===== Recommendation =====")
    print("Recommended destination:", best["name"])
    print("Type:", best["type"])
    print("Suggested minimum budget: PKR", best["min_budget"])
    print("Suggested days:", best["days"])
    print("Reason: This destination matches your available budget, travel time, and preferences.")

    print("\nOther possible options:")
    for score, place in possible[1:]:
        print("-", place["name"])
```
